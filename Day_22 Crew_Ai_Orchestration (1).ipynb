```python
# =========================================================
# AI Content Automation using CrewAI + Groq
# =========================================================

# -----------------------------
# 1. Install Libraries
# -----------------------------
!pip install -q crewai litellm

# -----------------------------
# 2. Imports
# -----------------------------
import os
from google.colab import userdata

from crewai import Agent, Task, Crew

# -----------------------------
# 3. Set Groq API Key
# -----------------------------
# Save your Groq API key in Colab Secrets as:
# gapi

os.environ["GROQ_API_KEY"] = userdata.get("gapi")

# -----------------------------
# 4. Create Research Agent
# -----------------------------
researcher = Agent(
    role="AI Researcher",

    goal="""
    Find latest trends and developments in
    Artificial Intelligence, Generative AI,
    LLMs, AI startups, and AI tools.
    """,

    backstory="""
    You are an expert AI researcher who tracks
    the latest innovations and breakthroughs
    in the AI industry.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# -----------------------------
# 5. Create Writer Agent
# -----------------------------
writer = Agent(
    role="LinkedIn Content Writer",

    goal="""
    Create engaging, professional, and viral
    LinkedIn posts from AI research summaries.
    """,

    backstory="""
    You are an expert content creator specializing
    in AI and technology-based social media content.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# -----------------------------
# 6. Create Research Task
# -----------------------------
research_task = Task(
    description="""
    Find 5 important latest AI trends or news topics.

    For each trend provide:
    - Title
    - Short summary
    - Why it matters
    """,

    expected_output="""
    A detailed list of 5 AI trends with explanations.
    """,

    agent=researcher
)

# -----------------------------
# 7. Create Writing Task
# -----------------------------
write_task = Task(
    description="""
    Using the AI research provided,
    create a professional LinkedIn post.

    The post should:
    - Be engaging
    - Professional
    - Easy to read
    - Include emojis
    - Include hashtags
    - Encourage audience interaction
    """,

    expected_output="""
    A complete LinkedIn post ready to publish.
    """,

    agent=writer,

    context=[research_task]
)

# -----------------------------
# 8. Create Crew
# -----------------------------
crew = Crew(
    agents=[researcher, writer],

    tasks=[research_task, write_task],

    verbose=True
)

# -----------------------------
# 9. Run CrewAI
# -----------------------------
result = crew.kickoff()

# -----------------------------
# 10. Print Final Output
# -----------------------------
print("\n================ FINAL OUTPUT ================\n")

print(result)
```

```python
# =========================================================
# Smart Job Application Assistant using CrewAI + Groq
# =========================================================

# ---------------------------------------------------------
# 1. Install Required Libraries
# ---------------------------------------------------------
!pip install -q crewai litellm

# ---------------------------------------------------------
# 2. Imports
# ---------------------------------------------------------
import os
from crewai import Agent, Task, Crew
from google.colab import userdata

# ---------------------------------------------------------
# 3. Set Groq API Key
# ---------------------------------------------------------
# Save your Groq API Key in Colab Secrets as:
# gapi

os.environ["GROQ_API_KEY"] = userdata.get("gapi")

# ---------------------------------------------------------
# 4. Create Agents
# ---------------------------------------------------------

# Agent 1: Job Analyzer
job_analyzer = Agent(
    role="Job Analyzer",

    goal="""
    Extract important skills, responsibilities,
    qualifications, and tools from job descriptions.
    """,

    backstory="""
    You are an HR expert skilled at analyzing job postings
    and identifying critical requirements for candidates.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# Agent 2: Resume Optimizer
resume_optimizer = Agent(
    role="Resume Optimizer",

    goal="""
    Improve candidate resumes by matching them
    with job requirements and highlighting relevant skills.
    """,

    backstory="""
    You are a professional career coach who helps students
    optimize resumes for better job selection chances.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# Agent 3: Cover Letter Writer
cover_letter_writer = Agent(
    role="Cover Letter Writer",

    goal="""
    Generate personalized and professional cover letters
    tailored to the target job role.
    """,

    backstory="""
    You are an expert writer who creates compelling
    cover letters for students and professionals.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# ---------------------------------------------------------
# 5. Input Data
# ---------------------------------------------------------

job_description = """
We are looking for a Data Analyst with experience in Python,
SQL, Power BI, Excel, and data visualization.

The candidate should be able to:
- Analyze datasets
- Create dashboards
- Communicate business insights
- Work with large amounts of data

Strong problem-solving and communication skills are preferred.
"""

candidate_resume = """
I am a BTech Computer Science student.

Skills:
- Python
- Basic SQL
- Power BI
- Excel

Projects:
- Created a Power BI sales dashboard
- Built a Python data analysis project

Certifications:
- Data Analytics Fundamentals
"""

# ---------------------------------------------------------
# 6. Task 1: Job Analysis
# ---------------------------------------------------------

analysis_task = Task(
    description=f"""
    Analyze the following job description and extract:

    - Required technical skills
    - Soft skills
    - Responsibilities
    - Tools and technologies
    - Important keywords

    Job Description:
    {job_description}
    """,

    expected_output="""
    Structured analysis of the job description
    with categorized skills and requirements.
    """,

    agent=job_analyzer
)

# ---------------------------------------------------------
# 7. Task 2: Resume Optimization
# ---------------------------------------------------------

resume_task = Task(
    description=f"""
    Compare the candidate resume with the job requirements.

    Tasks:
    - Identify missing skills
    - Suggest resume improvements
    - Rewrite resume professionally
    - Highlight relevant projects and strengths

    Candidate Resume:
    {candidate_resume}
    """,

    expected_output="""
    Optimized resume with improvement suggestions
    and better alignment to the job role.
    """,

    agent=resume_optimizer,

    context=[analysis_task]
)

# ---------------------------------------------------------
# 8. Task 3: Cover Letter Generation
# ---------------------------------------------------------

cover_letter_task = Task(
    description="""
    Create a professional and personalized cover letter using:

    - Job description
    - Candidate resume
    - Resume improvements

    The cover letter should:
    - Be professional
    - Show enthusiasm
    - Highlight relevant skills
    - Match the job role
    """,

    expected_output="""
    Final professional cover letter ready for submission.
    """,

    agent=cover_letter_writer,

    context=[analysis_task, resume_task]
)

# ---------------------------------------------------------
# 9. Create Crew
# ---------------------------------------------------------

crew = Crew(
    agents=[
        job_analyzer,
        resume_optimizer,
        cover_letter_writer
    ],

    tasks=[
        analysis_task,
        resume_task,
        cover_letter_task
    ],

    verbose=True
)

# ---------------------------------------------------------
# 10. Run CrewAI
# ---------------------------------------------------------

result = crew.kickoff()

# ---------------------------------------------------------
# 11. Print Final Output
# ---------------------------------------------------------

print("\n================ FINAL OUTPUT ================\n")

print(result)
```

```python
# =========================================================
# Automated Business Intelligence (BI) Report Generator
# using CrewAI + Groq
# =========================================================

# ---------------------------------------------------------
# 1. Install Required Libraries
# ---------------------------------------------------------
!pip install -q crewai litellm

# ---------------------------------------------------------
# 2. Imports
# ---------------------------------------------------------
import os
from crewai import Agent, Task, Crew
from google.colab import userdata

# ---------------------------------------------------------
# 3. Set Groq API Key
# ---------------------------------------------------------
# Save your Groq API key in Colab Secrets as:
# gapi

os.environ["GROQ_API_KEY"] = userdata.get("gapi")

# ---------------------------------------------------------
# 4. Create Agents
# ---------------------------------------------------------

# Agent 1: Data Analyst
data_analyst = Agent(
    role="Data Analyst",

    goal="""
    Analyze business and sales data to identify:
    - KPIs
    - Trends
    - Profitability
    - Performance patterns
    """,

    backstory="""
    You are a senior corporate data analyst who specializes
    in sales analysis, KPI tracking, and business intelligence.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# Agent 2: Business Strategist
business_strategist = Agent(
    role="Business Strategist",

    goal="""
    Convert business insights into actionable
    growth strategies and recommendations.
    """,

    backstory="""
    You are a business consultant who helps companies
    improve growth, profitability, and operational efficiency.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# Agent 3: Report Generator
report_generator = Agent(
    role="Executive Report Writer",

    goal="""
    Create professional executive-level business reports
    using insights and strategic recommendations.
    """,

    backstory="""
    You are an expert corporate report writer known for
    creating clear, structured, and professional reports.
    """,

    llm="groq/llama-3.3-70b-versatile",

    verbose=True
)

# ---------------------------------------------------------
# 5. Sample Business Data
# ---------------------------------------------------------

sales_data = """
Region: North | Revenue: 12000 | Cost: 8000
Region: South | Revenue: 15000 | Cost: 9000
Region: East  | Revenue: 10000 | Cost: 7000
Region: West  | Revenue: 18000 | Cost: 11000
"""

# ---------------------------------------------------------
# 6. Task 1: Data Analysis
# ---------------------------------------------------------

analysis_task = Task(
    description=f"""
    Analyze the following sales data and provide:

    - Total Revenue
    - Total Cost
    - Total Profit
    - Profit by region
    - Best performing region
    - Key business trends
    - Important KPIs
    - Observations and patterns

    Sales Data:
    {sales_data}
    """,

    expected_output="""
    Detailed business analysis with KPIs,
    profitability insights, and trends.
    """,

    agent=data_analyst
)

# ---------------------------------------------------------
# 7. Task 2: Strategy Recommendations
# ---------------------------------------------------------

strategy_task = Task(
    description="""
    Based on the sales analysis:

    - Suggest 3 strategic business improvements
    - Identify growth opportunities
    - Highlight business risks
    - Recommend cost optimization methods
    - Suggest future expansion ideas
    """,

    expected_output="""
    Strategic recommendations and business decisions
    based on the analyzed data.
    """,

    agent=business_strategist,

    context=[analysis_task]
)

# ---------------------------------------------------------
# 8. Task 3: Executive Report Generation
# ---------------------------------------------------------

report_task = Task(
    description="""
    Create a professional executive business report.

    The report must include:
    - Executive Summary
    - KPI Analysis
    - Revenue & Profit Insights
    - Key Trends
    - Strategic Recommendations
    - Risks and Opportunities
    - Final Conclusion

    Make the report:
    - Professional
    - Well-structured
    - Easy to understand
    - Suitable for company executives
    """,

    expected_output="""
    Final professional BI report ready for management review.
    """,

    agent=report_generator,

    context=[analysis_task, strategy_task]
)

# ---------------------------------------------------------
# 9. Create Crew
# ---------------------------------------------------------

crew = Crew(
    agents=[
        data_analyst,
        business_strategist,
        report_generator
    ],

    tasks=[
        analysis_task,
        strategy_task,
        report_task
    ],

    verbose=True
)

# ---------------------------------------------------------
# 10. Run CrewAI
# ---------------------------------------------------------

result = crew.kickoff()

# ---------------------------------------------------------
# 11. Print Final Output
# ---------------------------------------------------------

print("\n================ FINAL BUSINESS REPORT ================\n")

print(result)
```
